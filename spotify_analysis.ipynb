```python
import pandas as pd

import matplotlib.pyplot as plt

df = pd.read_csv("spotify-top-50-world.csv")

df.head()



df.info()



df.describe()

df.isnull().sum()

df.drop_duplicates(inplace=True)

top_artists = (

    df.groupby("artist")["popularity"]

      .mean()

      .sort_values(ascending=False)

      .head(10)

)



print(top_artists)

top_artists.plot(kind="bar", figsize=(10,5))



plt.title("Top 10 Most Popular Artists")

plt.xlabel("Artist")

plt.ylabel("Average Popularity")

plt.xticks(rotation=45)



plt.show()

top_songs = df.sort_values("popularity", ascending=False)



top_songs[["song", "artist", "popularity"]].head(10)

print(df["duration_ms"].mean())

df["is_explicit"].value_counts()

df["album_type"].value_counts()

df["artist"].value_counts().head(10)
```
